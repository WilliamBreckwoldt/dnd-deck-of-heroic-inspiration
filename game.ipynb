```python
import threading
import uvicorn
import time
from IPython.display import IFrame, display, HTML
from src.server import app

def start_server():
    # Runs FastAPI server quietly in the background
    uvicorn.run(app, host="0.0.0.0", port=47554, log_level="critical")

# Start thread
thread = threading.Thread(target=start_server, daemon=True)
thread.start()

# Wait a second for the server to spin up
time.sleep(1.5)

# Provide a direct clickable link in case the IFrame renders white due to Jupyter security settings
display(HTML("<h3><a href='http://127.0.0.1:47554' target='_blank'>➡️ Click here to open the game in a full browser tab</a></h3>"))

# Display the IFrame
# display(IFrame(src="http://127.0.0.1:47554", width="100%", height="1000px"))
```
